```python
import pandas as pd
import os

raw_path = "/content/project_folder/Financial-Credit-Risk-Analytics/01_data/raw"

inventory = []

for file in os.listdir(raw_path):

    if file.endswith(".xlsx"):

        file_path = os.path.join(raw_path, file)

        df = pd.read_excel(file_path)

        inventory.append({
            "File": file,
            "Rows": df.shape[0],
            "Columns": df.shape[1]
        })

        print("\n" + "="*80)
        print(file)
        print("Shape:", df.shape)
        print("Columns:")
        print(df.columns.tolist())
```

```python
import pandas as pd
import os

folder_path = "/content/project_folder/Financial-Credit-Risk-Analytics/01_data/raw"

dfs = []

for file in os.listdir(folder_path):
    if file.endswith(".xlsx"):
        df = pd.read_excel(os.path.join(folder_path, file))
        dfs.append(df)

financial_df = pd.concat(dfs, ignore_index=True)

print(financial_df.shape)
```

```python
clean_path = "/content/project_folder/Financial-Credit-Risk-Analytics/01_data/cleaned"

financial_df.to_csv(
    f"{clean_path}/financial_master_raw.csv",
    index=False
)
```
